## 2. Structured Programming (functions + control structures)






## Brief Explanation
It uses functions to organize logic

## Code Example
Complete the steps:

````python
def process():
    sum = 0

    for i in range(5):
        cal = float(input())
        sum += cal

    prom = sum / 5
    print(prom)

    if prom >= 70:
        print("Pass")
    else:
        print("Fail")

process()
````

## Your Modification
What did you change and why?
I wrapped the code inside a function called process(). I did this to properly apply structured programming, making the code more organized.


## Reflection
What did you learn from this approach and how does it compare to others?
I learned that structured programming makes code very clear and easy to follow. Compared to other complex methods, this approach is much faster.
